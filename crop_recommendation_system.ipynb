# Crop Recommendation System

This notebook describes the Kaggle crop recommendation dataset used by this project. The CLI and Flask API train from `crop_recommendation_dataset.csv`.

Dataset source: [Kaggle Crop Recommendation Dataset](https://www.kaggle.com/datasets/siddharthss/crop-recommendation-dataset?resource=download).

The CSV contains 2,200 rows, 22 crop labels, and seven input features: `N`, `P`, `K`, `temperature`, `humidity`, `ph`, and `rainfall`. Target: `label`.

Project contributors: Kavish Preet Singh (102483035), Ishan Satya Prakash (102303449), and Parul Garg (102303433).


```python
from pathlib import Path
import pandas as pd

data = pd.read_csv(Path("crop_recommendation_dataset.csv"))
data.head()
```

```python
expected = {"N", "P", "K", "temperature", "humidity", "ph", "rainfall", "label"}
print("Shape:", data.shape)
print("Columns match dataset:", set(data.columns) == expected)
print("Crop classes:", data["label"].nunique())
data["label"].value_counts().sort_index()
```

```python
data.groupby("label")[ ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"] ].mean().round(2)
```

Use the project CLI to train and save the shared model pipeline: `python cli.py train`. To make a prediction, use `python cli.py predict --interactive` or pass the seven feature values with `--values`.
